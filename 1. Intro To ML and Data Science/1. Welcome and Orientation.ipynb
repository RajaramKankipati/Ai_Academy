## What is Machine Learning, and what is Data Science?

**Data science** is the practice of extracting understanding and decisions from data:
collecting it, cleaning it, exploring it, testing hypotheses about it, and communicating
what you found. **Machine learning** is one tool inside that practice — a set of
algorithms that learn patterns from data automatically, rather than being explicitly
programmed with rules.

A useful mental model: every ML model is a function that maps inputs to outputs
($f: X \to y$), and "training" is the process of choosing that function's parameters
from data instead of writing them by hand. Everything from a simple linear regression
to a large language model fits this description — they differ enormously in *how*
$f$ is represented and *how* it is fit, not in the basic shape of the problem.

## A first taste: from raw numbers to an answer

Before the deep material, here is the entire shape of a data science question, answered
in a handful of cells, using the Gapminder dataset that reappears throughout this
course (`3. EDA and Visualization/gapminder.csv`).

**The question:** is there a relationship between a country's fertility rate and its
child mortality rate?

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../3. EDA and Visualization/gapminder.csv")
print(f"{len(df)} countries in this snapshot")
df[["fertility", "child_mortality", "life", "GDP"]].describe().round(2)
````

````python
plt.figure(figsize=(7, 5))
plt.scatter(df["fertility"], df["child_mortality"], alpha=0.6, color="steelblue")
plt.xlabel("fertility (children per woman)")
plt.ylabel("child mortality (per 1,000 live births)")
plt.title("Fertility vs child mortality, one point per country")
plt.show()

correlation = df["fertility"].corr(df["child_mortality"])
print(f"Correlation: {correlation:.3f}")
````

That is a strong positive relationship — and it is a real, well-documented one
(countries with higher child mortality tend to have higher fertility, partly because
families have more children to offset the risk that some will not survive). Notice what
this small example already touched on without spelling it out: **loading data**,
**visualizing** it before drawing conclusions, and **summarizing a relationship with a
single number**. Modules 3, 5 and 6 will each come back to exactly this pattern and go
much deeper — what makes a "strong" correlation, how to test whether it could be chance,
and whether it says anything about causation (it does not, on its own — see
`6. Statistical Foundations for Data Science/5. Correlation and Covariance.ipynb`).
