```python
s = "appearance"
t = "attitude"
```

```python
cost_of_insertion = 1
cost_of_deletion = 1
cost_of_substitution = 2
```

```python
def minimum_edit_distance(source, target):
    m, n = len(source), len(target)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    #Base cases
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    # fill the DP table
    for i in range(1, m+1): 
        for j in range(1, n+1):
            cost = 0 if source[i-1] == target[j-1] else cost_of_substitution
            dp[i][j] = min(dp[i-1][j] + cost_of_deletion , dp[i][j-1] + cost_of_insertion, dp[i-1][j-1] + cost)
    return dp[m][n]
```

```python
minimum_edit_distance(s, t)
```
