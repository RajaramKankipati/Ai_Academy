## 🟢 LEVEL 1: The Basics (Literals)

**Mission:** Match exact text strings.

A Regular Expression (Regex) is just a pattern. The simplest pattern is the text itself.

```python
import re

text = "The cat sat on the mat. And another cat is looking out"

# TODO: Change the pattern to find 'cat'
pattern = r"cat" 

match = re.search(pattern, text)
if match:
    print("✅ Found:", match.group())
else:
    print("❌ No match found. Try again!")
```

## 🟡 LEVEL 2: Wildcards & Character Sets

**Mission:** Match specific types of characters.

We can't always search for exact words. Sometimes we need to match *any digit* or *any letter*.

| Symbol | Meaning |
| :--- | :--- |
| `.` | Any character (except newline) |
| `\d` | Any Digit (0-9) |
| `\w` | Any Word Character (a-z, A-Z, 0-9, _) |
| `[abc]` | Only a, b, or c |
| `[a-z]` | Any range from a to z |

## 🔴 LEVEL 4: Anchors

**Mission:** Position matters!

*   `^` = Start of string
*   `$` = End of string

Without anchors, regex matches anywhere. With `^...$`, it must match the *whole* string.

```python
usernames = ["admin", "admin123", "_super_user_", "user name"]

# TODO: Enforce that the WHOLE string must be word characters
pattern = r"^_\w{5}_\w{4}_$"

print("--- Results ---")
for user in usernames:
    if re.search(pattern, user):
        print(f"✅ VALID: {user}")
    else:
        print(f"❌ INVALID: {user}")
```

## 🟣 LEVEL 5: Groups & Capturing (The Pro Tools)

**Mission:** Extract specific parts of the data using `( )`.

This is the most important part! We don't just want to know *if* it matches, we want to **extract** the data.

### 🧩 Challenge 1: Matching Emails
**Goal:** Extract the **Username** part of the email.
*   Ignore tags like `+regexone`

```python
pattern = r"([\w.]+)@hogwarts.com"
```

```python
re.findall(pattern, "tom.riddle@hogwarts.com")
```

```python
emails = [
    "tom@hogwarts.com", 
    "tom.riddle@hogwarts.com", 
    "tom.riddle+regexone@hogwarts.com", 
    "potter@hogwarts.com"
]

# TODO: Capture the username (Group 1)
pattern = r"^([\w.]+)(?:\+.*)?@"

print("--- Results ---")
for email in emails:
    match = re.search(pattern, email)
    if match:
        print(f"✅ CAPTURED: {match.group(1)}")
    else:
        print(f"❌ FAIL: {email}")
```

### 🧩 Challenge 2: Capturing HTML Tags
**Goal:** Extract the **Tag Name** (e.g., `div`, `a`).

```python
html_tags = [
    "<a>This is a link</a>",
    "<a href='https://regexone.com'>Link</a>",
    "<div class='test_style'>Test</div>",
    "<div>Hello <span>world</span></div>"
]

# TODO: Capture the tag name immediately after <
pattern = r"<(\w+)"

print("--- Results ---")
for tag in html_tags:
    match = re.search(pattern, tag)
    if match:
        print(f"✅ TAG: {match.group(1)}")
    else:
        print(f"❌ FAIL: {tag}")
```

### 🧩 Challenge 3: Capturing Filename Data
**Goal:** Extract **Name** and **Extension**, but only for images!

```python
files = [
    ".bash_profile", 
    "img0912.jpg", 
    "updated_img0912.png", 
    "documentation.html", 
    "favicon.gif"
]

# TODO: Capture name and extension (jpg, png, gif only)
pattern = r"^(\w+)\.(jpg|png|gif)$"

print("--- Results ---")
for f in files:
    match = re.search(pattern, f)
    if match:
        print(f"✅ FILE: {match.group(1)} | EXT: {match.group(2)}")
    else:
        print(f"❌ SKIP: {f}")
```

## ⚫️ LEVEL 6: Information Retrieval

**Scenario:** You have hacked into a server log. You need to extract specific IPs and Timestamps to find the intruder.

```python
log_data = """
[2024-01-08 10:00:01] User admin accessed from 192.168.1.5
2024/01/08 10:05 Error: Connection timeout
[2024-01-08 10:10:45] User root accessed from 10.0.0.1 (Suspicious)
[2024-01-08 10:11:00] User alice accessed from 192.168.1.12"""

print("--- Analysis Report ---")

# TODO: Extract Timestamp and IP for 192.168.x.x addresses
pattern = r"(\d{4}-\d{2}-\d{2}\s\d{2}:\d{2}:\d{2}).*?(192\.168\.\d+\.\d+)"
results = re.findall(pattern, log_data)

for timestamp, ip in results:
    print(f"🚨 Alert! Access at {timestamp} from {ip}")
```
