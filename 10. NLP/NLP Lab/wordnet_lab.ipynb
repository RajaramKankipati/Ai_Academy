# NLP Lab: Introduction to WordNet

## 1. What is WordNet and Synset?

### WordNet
WordNet is a large lexical database of English. Nouns, verbs, adjectives, and adverbs are grouped into sets of cognitive synonyms (synsets), each expressing a distinct concept. Synsets are interlinked by means of conceptual-semantic and lexical relations. using WordNet, you can find words, synonyms, antonyms, definitions, and examples.

### Synset
A **Synset**, or synonym set, is a collection of synonymous words that express the same concept. Some words have only one Synset and some have several depending on the context. For example, the word "bank" can refer to a financial institution or the side of a river. In WordNet, these would be in different synsets.

```python
# Importing WordNet
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.corpus import wordnet as wn
```

## 2. Hypernyms and Hyponyms

### Concepts
- **Hypernym**: A word with a broad meaning that more specific words fall under; a superordinate. For example, *color* is a hypernym of *red*.
- **Hyponym**: A word of more specific meaning than a general or superordinate term applicable to it. For example, *spoon* is a hyponym of *cutlery*.

These relationships organize meanings in a hierarchy.

```python
# Example of Hypernyms and Hyponyms
syn = wn.synsets('car')[0]  # Get the first synset for 'car'
print("Synset:", syn)

print("\nHypernyms:", syn.hypernyms())
print("Hyponyms:", syn.hyponyms())
```

## 3. Parts of Speech (POS) Example

Below are the POS tags for the requested words using WordNet Synsets.
Common POS tags in WordNet: `n` (noun), `v` (verb), `a` (adjective), `r` (adverb), `s` (satellite adjective).

a) Silently
b) Lamborghini
c) Beauty
d) Maldives

```python
words = ["silently", "Lamborghini", "beauty", "Maldives"]

for word in words:
    synsets = wn.synsets(word)
    print(f"\nWord: {word}")
    if synsets:
        # Print the POS of the first synset found
        print(f"First Synset: {synsets[0].name()}")
        print(f"POS: {synsets[0].pos()}")
        print(f"Definition: {synsets[0].definition()}")
    else:
        print("No synsets found in WordNet")
```

## 4. Find Synonyms and Antonyms using NLTK WordNet

We will find synonyms and antonyms for the word **"bad"**.

```python
synonyms = []
antonyms = []

target_word = "bad"

for syn in wn.synsets(target_word):
    for lemma in syn.lemmas():
        # Add synonym
        synonyms.append(lemma.name())
        # Add antonym if available
        if lemma.antonyms():
            antonyms.append(lemma.antonyms()[0].name())

# Using set to remove duplicates
print(f"Synonyms for '{target_word}':", set(synonyms))
print(f"Antonyms for '{target_word}':", set(antonyms))
```

## 5. Lexical Relations

Lexical relations describe how words are related to one another in the lexicon.

### Common Lexical Relations:

1.  **Synonymy**: Words with similar meanings (e.g., *happy*, *joyful*).
2.  **Antonymy**: Words with opposite meanings (e.g., *hot*, *cold*).
3.  **Hypernymy**: Generalization (e.g., *furniture* is a hypernym of *chair*).
4.  **Hyponymy**: Specialization (e.g., *chair* is a hyponym of *furniture*).
5.  **Meronymy**: Part-whole relationship (e.g., *wheel* is a meronym of *car*).
6.  **Holonymy**: Whole-part relationship (e.g., *car* is a holonym of *wheel*).

WordNet organizes these relations systematically.

```python
# Examples of Meronyms (parts) and Holonyms (wholes)
tree = wn.synset('tree.n.01')

print("Tree parts (Meronyms):", tree.part_meronyms())
print("Tree substance (Substance Meronyms):", tree.substance_meronyms())

atom = wn.synset('atom.n.01')
print("\nAtom part of (Holonyms):", atom.part_holonyms())
```

## 6. Industrial Applications

In real-world NLP problems, WordNet is often used for tasks that require an understanding of word meanings and relationships beyond simple string matching. Some common applications include **Semantic Similarity**, **Query Expansion**, and **Chatbot Intent Recognition**.

### A. Semantic Similarity

Semantic similarity measures how close two words are in terms of their meaning. This is crucial for:
-   **Search Engines**: To understand that a search for "automobile" matches documents containing "car".
-   **Recommender Systems**: To recommend items similar to what a user likes.
-   **Plagiarism Detection**: To find paraphrased content.

One common metric is the **Wu-Palmer Similarity**, which calculates similarity based on the depth of the two synsets in the taxonomy and that of their Least Common Subsumer (most specific ancestor node).

```python
# Semantic Similarity Example

dog = wn.synset('dog.n.01')
cat = wn.synset('cat.n.01')
car = wn.synset('car.n.01')

# Wu-Palmer Similarity
dog_cat_sim = dog.wup_similarity(cat)
dog_car_sim = dog.wup_similarity(car)

print(f"Similarity between Dog and Cat: {dog_cat_sim:.2f}")
print(f"Similarity between Dog and Car: {dog_car_sim:.2f}")

# Interpretation: Dog and Cat are much more similar (both animals) than Dog and Car.
```

### B. Query Expansion

Query expansion is a technique used in information retrieval to increase the recall of a search engine. It involves adding synonyms or related terms to the original user query.

**Problem**: A user searches for "cheap laptop", but a relevant document only contains "affordable notebook". Standard keyword search might miss this.
**Solution**: Use WordNet to expand "cheap" to include "inexpensive", "affordable" and "laptop" to include "notebook".

```python
def expand_query(query_str):
    expanded_terms = set()
    words = query_str.split()
    
    for word in words:
        expanded_terms.add(word) # Add original word
        # Find synonyms in WordNet
        for syn in wn.synsets(word):
            for lemma in syn.lemmas():
                # Avoid multi-word expressions for simplicity in this example
                if "_" not in lemma.name():
                    expanded_terms.add(lemma.name())
    
    return list(expanded_terms)

original_query = "cheap laptop"
expanded_query = expand_query(original_query)

print(f"Original Query: {original_query}")
print(f"Expanded Query: {expanded_query}")
```

### C. Chatbot Intent Recognition (Real-World Application)

In the industry, conversational AI (chatbots) often struggle with the variety of ways humans express the same request. By leveraging WordNet, a chatbot can understand that strings of text containing words like *"purchase"*, *"acquire"*, or *"get"* all map to the same **"buy"** intent without having to hardcode every possible synonym.

```python
def check_intent(user_input, target_verb):
    # Get synonyms for the target verb (e.g., 'buy')
    target_synonyms = set()
    for syn in wn.synsets(target_verb, pos=wn.VERB):
        for lemma in syn.lemmas():
            target_synonyms.add(lemma.name())
            
    # Check if any word in the user input matches the target's synonyms
    words = user_input.lower().replace("?", "").split()
    for word in words:
        if word in target_synonyms:
            return True
    return False

# Target action for an e-commerce bot
intent = "buy"

# Different user inputs that mean the same thing
inputs = [
    "I want to purchase a new laptop",
    "Where can I acquire a phone?",
    "I need to return this item" # Different intent
]

print("Chatbot Intent Matching:\n")
for text in inputs:
    if check_intent(text, intent):
        print(f"User: '{text}' -> Bot mapped to intent: {intent.upper()}")
    else:
        print(f"User: '{text}' -> Bot mapped to intent: UNKNOWN (needs human agent)")
```
