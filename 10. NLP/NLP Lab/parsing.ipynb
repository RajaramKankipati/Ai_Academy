## 1. What is a Parser?

### Definition

A parser is a component in Natural Language Processing (NLP) that analyzes the grammatical structure of a sentence according to a formal grammar. It takes a sequence of tokens (words) as input and produces a structured representation, typically in the form of a parse tree or dependency graph.

Parsing helps determine:

* Syntactic structure of sentences
* Relationships between words
* Phrase groupings
* Grammatical correctness

### Why Parsing is Important in NLP

* Machine Translation
* Question Answering Systems
* Information Extraction
* Sentiment Analysis
* Speech Recognition


## Types of Parsers

Parsers in NLP can be broadly classified as follows:

### 1. Top-Down Parsers

* Start from the root (start symbol) and attempt to rewrite it to the input sentence.
* Build the parse tree from root to leaves.
* Example: Recursive Descent Parser.

### 2. Bottom-Up Parsers

* Start from the input tokens and attempt to build up to the start symbol.
* Construct the parse tree from leaves to root.
* Example: Shift-Reduce Parser.

### 3. Constituency Parsers

* Represent sentence structure as nested phrase structures.
* Output is a phrase structure tree.

### 4. Dependency Parsers

* Represent grammatical relations between words.
* Output is a dependency graph.

### 5. Chart Parsers

* Use dynamic programming to store intermediate results.
* Efficient for ambiguous grammars.
* Example: CYK Parser, Earley Parser.



## 2. Recursive Descent Parser

### Definition

A Recursive Descent Parser is a top-down parsing technique that uses a set of recursive procedures to process the input according to grammar rules.

Each non-terminal in the grammar corresponds to a function. The parser attempts to match the input string by recursively expanding grammar productions.

### Key Characteristics

* Top-down approach
* Backtracking may be required
* Simple to implement
* Works well with LL(1) grammars

### Working Principle

1. Start from the start symbol.
2. Apply production rules recursively.
3. Match terminals with input tokens.
4. If mismatch occurs, backtrack (if allowed).

### Advantages

* Easy to understand and implement
* Good for simple grammars

### Limitations

* Inefficient with backtracking
* Cannot handle left-recursive grammars directly



## 3. Shift-Reduce Parser

### Definition

A Shift-Reduce Parser is a bottom-up parsing technique that constructs a parse tree by shifting input symbols onto a stack and reducing them using grammar rules.

It attempts to reduce the input string to the start symbol by reversing rightmost derivations.

### Core Operations

1. Shift – Move the next input token onto the stack.
2. Reduce – Replace a sequence of stack symbols with a non-terminal according to a grammar rule.
3. Accept – Successful parsing when input is consumed and stack contains only the start symbol.
4. Reject – Parsing fails.

### Components

* Input buffer
* Stack
* Parsing table (in advanced versions like LR parsers)

### Types of Shift-Reduce Parsers

* LR(0)
* SLR(1)
* LALR(1)
* Canonical LR(1)

### Advantages

* More powerful than simple top-down parsers
* Can handle a larger class of grammars
* Efficient for deterministic grammars

### Limitations

* More complex to implement
* Requires construction of parsing tables

### Grammar rules
S  → NP VP
NP → Det N
NP → N
VP → V NP
VP → V
Det → the | a
N  → boy | telescope
V  → saw

```python
### Recursive Descent Parser (Top-Down with Backtracking)

### Grammar definition
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"], ["N"]],
    "VP": [["V", "NP"], ["V"]],
    "Det": [["the"], ["a"]],
    "N": [["boy"], ["telescope"]],
    "V": [["saw"]]
}
```

```python

class RecursiveDescentParser:
    def __init__(self, grammar, tokens):
        self.grammar = grammar
        self.tokens = tokens
        self.position = 0

    def parse(self):
        success = self.parse_non_terminal("S")
        return success and self.position == len(self.tokens)

    def parse_non_terminal(self, symbol):
        print("Processing symbol:", symbol)
        if symbol not in self.grammar:
            return self.match(symbol)

        print("Productions for symbol:", self.grammar[symbol])
        for production in self.grammar[symbol]:
            saved_position = self.position
            success = True

            for sub_symbol in production:
                if not self.parse_non_terminal(sub_symbol):
                    success = False
                    break
            
            print("Backtracking to position:", saved_position)
            if success:
                return True
            else:
                self.position = saved_position  # Backtrack
            
        return False

    def match(self, terminal):
        if self.position < len(self.tokens) and self.tokens[self.position] == terminal:
            self.position += 1
            return True
        return False
```

```python

# Example input
sentence = "the boy saw telescope"
tokens = sentence.split()

parser = RecursiveDescentParser(grammar, tokens)
```

```python
if parser.parse():
    print("Sentence is syntactically correct.")
else:
    print("Sentence is syntactically incorrect.")
```

#### How It Works

Start with goal S

Expand using grammar rules

Recursively process non-terminals

Match terminals with input words

If mismatch occurs → backtrack

Accept if all words are consumed successfully

## Shift Reduce Parser

### Shift Reduce Parser

#### Grammar definition
S  → NP VP
NP → Det N
NP → N
VP → V NP
VP → V
Det → the | a
N  → boy | telescope
V  → saw

Bottom-Up Parsing Strategy
Operations Used

Shift → Move next word to stack

Reduce → If stack top matches RHS of a production, replace with LHS

Accept → If stack contains only S and input is empty

Reject → Otherwise

```python
# Shift-Reduce Parser (Bottom-Up Parsing)

# Grammar rules (LHS -> RHS)
grammar = [
    ("S", ["NP", "VP"]),
    ("NP", ["Det", "N"]),
    ("NP", ["N"]),
    ("VP", ["V", "NP"]),
    ("VP", ["V"]),
    ("Det", ["the"]),
    ("Det", ["a"]),
    ("N", ["boy"]),
    ("N", ["telescope"]),
    ("V", ["saw"])
]

class ShiftReduceParser:
    def __init__(self, grammar, tokens):
        self.grammar = grammar
        self.tokens = tokens
        self.stack = []

    def reduce(self):
        for lhs, rhs in self.grammar:
            if len(self.stack) >= len(rhs):
                if self.stack[-len(rhs):] == rhs:
                    # Perform reduction
                    self.stack = self.stack[:-len(rhs)]
                    self.stack.append(lhs)
                    return True
        return False

    def parse(self):
        input_buffer = self.tokens[:]

        while True:
            # Try reduction first
            if self.reduce():
                continue

            # Shift if input remains
            if input_buffer:
                self.stack.append(input_buffer.pop(0))
            else:
                break

        return self.stack == ["S"]


# Example input
sentence = "the boy saw telescope"
tokens = sentence.split()

parser = ShiftReduceParser(grammar, tokens)

if parser.parse():
    print("Sentence is syntactically correct.")
else:
    print("Sentence is syntactically incorrect.")
```
